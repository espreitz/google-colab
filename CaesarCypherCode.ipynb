```python
try:
    choice = int(input("Enter 1 to encrypt or 2 to decrypt: "))
except ValueError:
    print("Error: Please enter either 1 or 2.")


letters = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']
user_message = list(str(input("Input the message you would like to encrypt\n")))
shift = int(input("Input the number (positive or negative) of letters you would like to shift\n"))


def encrypt_message(user_message1, shift1):
    encrypted_message = []
    encrypted_letter = ""
    print(user_message1)
    for letter in user_message1:
        if letter in user_message1:
            x = letters.index(letter)
            x = (x + shift1) % 26
            encrypted_letter = letters[x]
            encrypted_message.append(encrypted_letter)
    return ''.join(encrypted_message)

def decrypt_message(user_message1, shift1):
    decrypted_message = []
    decrypted_letter = ""
    print(user_message1)
    for letter in user_message1:
        if letter in letters:
            x = letters.index(letter)
            x = (x - shift1) % 26
            decrypted_message.append(letters[x])
            decrypted_message.append(decrypted_letter)
    return ''.join(decrypted_message)


if choice == 1:
    print("Encrypted message:", encrypt_message(user_message, shift))

elif choice == 2:
    print("Decrypted message:", decrypt_message(user_message, shift))
```
